```python
import pandas as pd 
import requests
import os
import json
import numpy as np
```

```python
# Opening JSON file
f = open(r'C:\Users\Alex\Documents\NHL-Stats\2024-2025\Data\JSON_files\game_story_all.json')
game_data = json.load(f)
```

```python
#from IPython.display import JSON
#JSON(game_data)
```

```python
# create lists for data frame 
game_num = []
game_date = []
venue = []
venue_location = []

a_team_id = []
h_team_id = []
a_team_code = []
h_team_code = []
a_goals = []
h_goals = []
a_sog = []
h_sog = []
a_fo_win_perc = []
h_fo_win_perc = []
a_pp = []
h_pp = []
a_pp_perc = []
h_pp_perc = []
a_pim = []
h_pim = []
a_hits = []
h_hits = []
a_blocked_shots = []
h_blocked_shots = []
a_giveaways = []
h_giveaways = []
a_takeaways = []
h_takeaways = []

#star_1_num = []
#star_1_id = []
#star_2_num = []
#star_2_id = []
#star_3_num = []
#star_3_id = []
```

```python
#pull data from JSON into a list\
for i in range(len(game_data)):
    game_num.append(game_data[i]['id'])
    game_date.append(game_data[i]['gameDate'])
    venue.append(game_data[i]['venue']['default'])
    venue_location.append(game_data[i]['venueLocation']['default'])
    a_team_id.append(game_data[i]['awayTeam']['id'])
    h_team_id.append(game_data[i]['homeTeam']['id'])
    a_team_code.append(game_data[i]['awayTeam']['abbrev'])
    h_team_code.append(game_data[i]['homeTeam']['abbrev'])
    a_goals.append(game_data[i]['awayTeam']['score'])
    h_goals.append(game_data[i]['homeTeam']['score'])
    a_sog.append(game_data[i]['summary']['teamGameStats'][0]['awayValue'])
    h_sog.append(game_data[i]['summary']['teamGameStats'][0]['homeValue'])
    a_fo_win_perc.append(game_data[i]['summary']['teamGameStats'][1]['awayValue'])
    h_fo_win_perc.append(game_data[i]['summary']['teamGameStats'][1]['homeValue'])
    a_pp.append(game_data[i]['summary']['teamGameStats'][2]['awayValue'])
    h_pp.append(game_data[i]['summary']['teamGameStats'][2]['homeValue'])
    a_pp_perc.append(game_data[i]['summary']['teamGameStats'][3]['awayValue'])
    h_pp_perc.append(game_data[i]['summary']['teamGameStats'][3]['homeValue'])
    a_pim.append(game_data[i]['summary']['teamGameStats'][4]['awayValue'])
    h_pim.append(game_data[i]['summary']['teamGameStats'][4]['homeValue'])
    a_hits.append(game_data[i]['summary']['teamGameStats'][5]['awayValue'])
    h_hits.append(game_data[i]['summary']['teamGameStats'][5]['homeValue'])
    a_blocked_shots.append(game_data[i]['summary']['teamGameStats'][6]['awayValue'])
    h_blocked_shots.append(game_data[i]['summary']['teamGameStats'][6]['homeValue'])
    a_giveaways.append(game_data[i]['summary']['teamGameStats'][7]['awayValue'])
    h_giveaways.append(game_data[i]['summary']['teamGameStats'][7]['homeValue'])
    a_takeaways.append(game_data[i]['summary']['teamGameStats'][8]['awayValue'])
    h_takeaways.append(game_data[i]['summary']['teamGameStats'][8]['homeValue'])  
```

```python
#create a data frame with the 
game_story_data = pd.DataFrame(list(zip(game_num, game_date, venue, venue_location, a_team_id, h_team_id,
                            a_team_code, h_team_code, a_goals, h_goals, a_sog, h_sog, a_fo_win_perc, h_fo_win_perc,
                            a_pp, h_pp, a_pp_perc, h_pp_perc, a_pim, h_pim, a_hits, h_hits, a_blocked_shots,
                            h_blocked_shots, a_giveaways, h_giveaways, a_takeaways, h_takeaways)),
               columns =['game_num', 'game_date', 'venue', 'venue_location', 'a_team_id', 'h_team_id',
                            'a_team_code', 'h_team_code', 'a_goals', 'h_goals', 'a_sog', 'h_sog', 'a_fo_win_perc', 'h_fo_win_perc',
                            'a_pp', 'h_pp', 'a_pp_perc', 'h_pp_perc', 'a_pim', 'h_pim', 'a_hits', 'h_hits', 'a_blocked_shots',
                            'h_blocked_shots', 'a_giveaways', 'h_giveaways', 'a_takeaways', 'h_takeaways'])
```

```python
game_story_data.head()
```

#### Cleaning pp data so its ppg

```python
# remove anything after / - first number is the number of ppg
game_story_data['a_pp'] = game_story_data['a_pp'].str.split('/').str[0]
game_story_data['h_pp'] = game_story_data['h_pp'].str.split('/').str[0]
```

```python
#remame pp_column
game_story_data.rename(columns={'a_pp': 'a_ppg', 'h_pp': 'h_ppg'}, inplace=True)
```
