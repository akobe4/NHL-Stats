```python
import pandas as pd 
import requests
import os
import json
import numpy as np
```

```python
# Opening JSON file
f = open(r'C:\Users\Alex\Documents\NHL-Stats\2024-2025\Data\JSON_files\glossary_info.json')
glossary_data = json.load(f)
```

```python
#from IPython.display import JSON
#JSON(glossary_data)
```

```python
#create empty lists for the dataframe
def_id = []
abbrev = []
name = []
definition = []
first_season_for_stat = []
last_updated = []
```

```python
for i in range(len(glossary_data['data'])):
    def_id.append(glossary_data['data'][i]['id'])
    abbrev.append(glossary_data['data'][i]['abbreviation'])
    name.append(glossary_data['data'][i]['fullName'])
    definition.append(glossary_data['data'][i]['definition'])
    first_season_for_stat.append(glossary_data['data'][i]['firstSeasonForStat'])
    last_updated.append(glossary_data['data'][i]['lastUpdated'])
```

```python
#create a data frame with faceoff information 
glossary = pd.DataFrame(list(zip(def_id, abbrev, name, definition, first_season_for_stat, last_updated)),
               columns =['def_id', 'abbrev', 'name', 'definition', 'first_season_for_stat', 'last_updated'])
```
