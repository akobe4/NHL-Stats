```python
import pandas as pd 
import requests
import os
import json
import numpy as np
```

```python
#create game number list 
#game numbers for the regular season range from 0001-1312
game_number_list = list(range(1,1313))
```

```python
#pad the game number list with leading zeros - creates a 4 digit number 
game_number_list_pad = [str(number).zfill(4) for number in game_number_list]
```

# Pull all Play-by-Play Data

```python
#create the url's for each game number 
all_url = []

for i in range(len(game_number_list_pad)):
    game_number = game_number_list_pad[i]
    game_number_url = 'https://api-web.nhle.com/v1/gamecenter/202502'+str(game_number)+'/play-by-play'
    all_url.append(game_number_url)
```

```python
#Pull the play-by-play data from the NHL 
game_data_all = []

for i in range(len(all_url)):
    response = requests.get(all_url[i])
    game_data_all.append(response.json())
```
