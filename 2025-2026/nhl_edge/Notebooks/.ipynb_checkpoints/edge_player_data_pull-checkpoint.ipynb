```python
#import player info file for player id's
skater_id_data = pd.read_csv(r'C:\Users\Alex\Documents\NHL-Stats\2025-2026\regular_season\Data\csv\player_info.csv')
```

```python
#create list of playerid's 
skater_id_list = skater_id_data['player_id'].tolist()
```

# Skater Data

### Skater Detail 

###### Retrieve player rankings for NHL Edge data, Includes top shot speed, skating speed, distance skated, shot on goal summary/details, zone time percentages.

```python
#create url's for each player id
all_url = []

for i in range(len(skater_id_list)):
    skater_id = skater_id_list[i]
    skater_id_url = 'https://api-web.nhle.com/v1/edge/skater-detail/'+str(skater_id)+'/20252026/2'
    all_url.append(skater_id_url)
```

```python
skater_details = []

for i in range(len(all_url)):
    response = requests.get(all_url[i])
    skater_details.append(response.json())
```
