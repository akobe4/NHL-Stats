## Skater Shot Speed Details 

```python
# Opening JSON file
f = open(r'C:\Users\Alex\Documents\NHL-Stats\2025-2026\nhl_edge\Data\JSONs\Skaters\skater_shot_speed_detail.json')
shot_speed_data = json.load(f)
```

```python
#from IPython.display import JSON
#JSON(shot_speed_data)
```

```python
#create empty lists for the data 
first_name = []
last_name = []
max_speed_imp = []
max_speed_met = []
avg_speed_imp = []
avg_speed_met = []
shots_over_100 = []
shots_90_to_100 = []
shots_80_to_90  = []
shots_70_to_80 = []
```

```python
#pull the data from json to a list 
for i in range(len(shot_speed_data)):    
    first_name.append(shot_speed_data[i]['shotSpeedDetails']['topShotSpeed']['overlay']['player']['firstName']['default'])
    last_name.append(shot_speed_data[i]['shotSpeedDetails']['topShotSpeed']['overlay']['player']['lastName']['default'])
    max_speed_imp.append(shot_speed_data[i]['shotSpeedDetails']['topShotSpeed']['imperial'])
    max_speed_met.append(shot_speed_data[i]['shotSpeedDetails']['topShotSpeed']['metric'])
    avg_speed_imp.append(shot_speed_data[i]['shotSpeedDetails']['avgShotSpeed']['imperial'])
    avg_speed_met.append(shot_speed_data[i]['shotSpeedDetails']['avgShotSpeed']['metric'])
    shots_over_100.append(shot_speed_data[i]['shotSpeedDetails']['shotAttemptsOver100']['value'])
    shots_90_to_100.append(shot_speed_data[i]['shotSpeedDetails']['shotAttempts90To100']['value'])
    shots_80_to_90.append(shot_speed_data[i]['shotSpeedDetails']['shotAttempts80To90']['value'])
    shots_70_to_80.append(shot_speed_data[i]['shotSpeedDetails']['shotAttempts70To80']['value'])
```

```python
#create a data frame with shot speed details
shot_speed_data = pd.DataFrame(list(zip(first_name, last_name, max_speed_imp, max_speed_met, avg_speed_imp, avg_speed_met,
                                            shots_over_100, shots_90_to_100, shots_80_to_90, shots_70_to_80)),
               columns =['first_name', 'last_name', 'max_speed_imp', 'max_speed_met', 'avg_speed_imp', 'avg_speed_met',
                                            'shots_over_100', 'shots_90_to_100', 'shots_80_to_90', 'shots_70_to_80'])
```
