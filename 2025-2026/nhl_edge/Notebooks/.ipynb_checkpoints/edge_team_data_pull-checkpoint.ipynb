```python
import pandas as pd 
import requests
import os
import json
import numpy as np
```

```python
#import team data file for team id's 
team_id_data = pd.read_csv(r'C:\Users\Alex\Documents\NHL-Stats\2025-2026\regular_season\Data\csv\teams.csv')
```

```python
#create list of team id's 
team_id_list = team_id_data['team_id'].tolist()
```

## Team Data

### Team Details

###### Retrieve team-based ranking for NHL Edge data

```python
#create url's for each team id
all_url = []

for i in range(len(team_id_list)):
    team_id = team_id_list[i]
    team_id_url = 'https://api-web.nhle.com/v1/edge/team-detail/'+str(team_id)+'/20252026/2'
    all_url.append(team_id_url)
```

```python
team_details_all = []

for i in range(len(all_url)):
    response = requests.get(all_url[i])
    team_details_all.append(response.json())
```

```python
#from IPython.display import JSON
#JSON(team_details_all)
```
