## Goalie Detail

```python
# Opening JSON file
f = open(r'C:\Users\Alex\Documents\NHL-Stats\2025-2026\nhl_edge\Data\JSONs\Goalies\goalie_detail.json')
goalie_detail_data = json.load(f)
```

```python
#from IPython.display import JSON
#JSON(goalie_detail_data)
```

```python
#create empty lists for goalie details data 
goalie_id = []
wins = [] 
losses = []
ot_losses = []
games_played = []
save_pctg = []
gaa = []
gaa_percentile = []
goal_support = []
goal_support_percentile = []
point_pctg = []
point_pctg_percentile = []
goal_diff_per_60 = []
goal_diff_per_60_percentile = []
```

```python
#pull data from JSONs into lists 
for i in range(len(goalie_detail_data)):
    goalie_id.append(goalie_detail_data[i]['player']['id'])
    wins.append(goalie_detail_data[i]['player']['wins'])
    losses.append(goalie_detail_data[i]['player']['losses'])
    ot_losses.append(goalie_detail_data[i]['player']['overtimeLosses'])
    games_played.append(goalie_detail_data[i]['player']['gamesPlayed'])
    save_pctg.append(goalie_detail_data[i]['player']['savePctg'])
    gaa.append(goalie_detail_data[i]['player']['goalsAgainstAvg'])
    gaa_percentile.append(goalie_detail_data[i]['stats']['goalsAgainstAvg']['percentile'])
    goal_support.append(goalie_detail_data[i]['stats']['goalSupportAvg']['value'])
    goal_support_percentile.append(goalie_detail_data[i]['stats']['goalSupportAvg']['percentile'])
    point_pctg.append(goalie_detail_data[i]['stats']['pointPctg']['value'])
    point_pctg_percentile.append(goalie_detail_data[i]['stats']['pointPctg']['percentile'])
    goal_diff_per_60.append(goalie_detail_data[i]['stats']['goalDifferentialPer60']['value'])
    goal_diff_per_60_percentile.append(goalie_detail_data[i]['stats']['goalDifferentialPer60']['percentile'])    
```

```python
#create a data frame with goalie details data 
goalie_detail_data = pd.DataFrame(list(zip(goalie_id, wins, losses, ot_losses, games_played, save_pctg, gaa, gaa_percentile, goal_support,
                                                goal_support_percentile, point_pctg, point_pctg_percentile, goal_diff_per_60, goal_diff_per_60_percentile)),
               columns =['goalie_id', 'wins', 'losses', 'ot_losses', 'games_played', 'save_pctg', 'gaa', 'gaa_percentile', 'goal_support',
                                                'goal_support_percentile', 'point_pctg', 'point_pctg_percentile', 'goal_diff_per_60', 'goal_diff_per_60_percentile'])
```

### Goalie 5v5 Detail

```python
# Opening JSON file
f = open(r'C:\Users\Alex\Documents\NHL-Stats\2025-2026\nhl_edge\Data\JSONs\Goalies\goalie_5v5_detail.json')
goalie_5v5_data = json.load(f)
```

```python
#from IPython.display import JSON
#JSON(goalie_5v5_data)
```

```python
#create lists for goalie 5v5 details 
save_pctg = []
save_pctg_percentile = []
save_pctg_close = []
save_pctg_close_percentile = []
shots = []
shots_percentile = []
shots_per_60 = []
shots_per_60_percentile = []
```

```python
#pull data from JSONs into lists 
for i in range(len(goalie_5v5_data)):
    try:
        save_pctg.append(goalie_5v5_data[i]['savePctg5v5Details']['savePctg']['value'])
        save_pctg_percentile.append(goalie_5v5_data[i]['savePctg5v5Details']['savePctg']['percentile'])
        save_pctg_close.append(goalie_5v5_data[i]['savePctg5v5Details']['savePctgClose']['value'])
        save_pctg_close_percentile.append(goalie_5v5_data[i]['savePctg5v5Details']['savePctgClose']['percentile'])
        shots.append(goalie_5v5_data[i]['savePctg5v5Details']['shots']['value'])
        shots_percentile.append(goalie_5v5_data[i]['savePctg5v5Details']['shots']['percentile'])
        shots_per_60.append(goalie_5v5_data[i]['savePctg5v5Details']['shotsPer60']['value'])
        shots_per_60_percentile.append(goalie_5v5_data[i]['savePctg5v5Details']['shotsPer60']['percentile'])

    except KeyError:
        save_pctg.append('NaN')
        save_pctg_percentile.append('NaN')
        save_pctg_close.append('NaN')
        save_pctg_close_percentile.append('NaN')
        shots.append('NaN')
        shots_percentile.append('NaN')
        shots_per_60.append('NaN')
        shots_per_60_percentile.append('NaN')      
```

```python
#create a data frame with mgoalie 5v5 details 
goalie_5v5_details = pd.DataFrame(list(zip(goalie_id_list, save_pctg, save_pctg_percentile, save_pctg_close, save_pctg_close_percentile,
                                            shots, shots_percentile, shots_per_60, shots_per_60_percentile)),
               columns =['goalie_id', 'save_pctg', 'save_pctg_percentile', 'save_pctg_close', 'save_pctg_close_percentile',
                                            'shots', 'shots_percentile', 'shots_per_60', 'shots_per_60_percentile'])
```
