### Max Skating Speed Burst Data 

```python
#create lists for max speed bursts data 
skater_id = []
skater_team = []
max_speed_imp = [] 
max_speed_met = []
max_speed_percentile = []
game_date = []
away_team = []
home_team = []
away_team_score = []
home_team_score = []
period = []
time_in_period = []
```

```python
#pull the data from JSONs into lists 
for i in range(len(skater_detail_data)):
    skater_id.append(skater_detail_data[i]['player']['id'])
    skater_team.append(skater_detail_data[i]['player']['team']['abbrev'])
    max_speed_imp.append(skater_detail_data[i]['skatingSpeed']['speedMax']['imperial']) 
    max_speed_met.append(skater_detail_data[i]['skatingSpeed']['speedMax']['metric'])
    max_speed_percentile.append(skater_detail_data[i]['skatingSpeed']['speedMax']['percentile'])
    game_date.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['gameDate'])
    away_team.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['awayTeam']['abbrev'])
    home_team.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['homeTeam']['abbrev'])
    away_team_score.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['awayTeam']['score'])
    home_team_score.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['homeTeam']['score'])
    period.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['periodDescriptor']['number'])
    time_in_period.append(skater_detail_data[i]['skatingSpeed']['speedMax']['overlay']['timeInPeriod']) 
```

```python
#create a data frame with max skating speed burst data 
max_skating_speed_burst = pd.DataFrame(list(zip(skater_id, skater_team, max_speed_imp, max_speed_met, max_speed_percentile, game_date,
                                                    away_team, home_team, away_team_score, home_team_score, period, time_in_period)),
               columns =['skater_id', 'skater_team', 'max_speed_imp', 'max_speed_met', 'max_speed_percentile', 'game_date',
                                            'away_team', 'home_team', 'away_team_score', 'home_team_score', 'period', 'time_in_period'])
```

### Max Shot Speed Data 

```python
#create lists for max shot data 
skater_id = []
shooter_team = []
max_speed_imp = []
max_speed_met = []
max_speed_percentile = []
game_date = []
away_team = []
home_team = []
away_team_score = []
home_team_score = []
period = []
time_in_period = []
```

```python
#pull data from JSONs into lists 
for i in range(len(skater_detail_data)):
    skater_id.append(skater_detail_data[i]['player']['id'])
    shooter_team.append(skater_detail_data[i]['player']['team']['abbrev'])

    try:
        max_speed_imp.append(skater_detail_data[i]['topShotSpeed']['imperial'])
        max_speed_met.append(skater_detail_data[i]['topShotSpeed']['metric'])
        max_speed_percentile.append(skater_detail_data[i]['topShotSpeed']['percentile'])
        game_date.append(skater_detail_data[i]['topShotSpeed']['overlay']['gameDate'])
        away_team.append(skater_detail_data[i]['topShotSpeed']['overlay']['awayTeam']['abbrev'])
        home_team.append(skater_detail_data[i]['topShotSpeed']['overlay']['homeTeam']['abbrev'])
        away_team_score.append(skater_detail_data[i]['topShotSpeed']['overlay']['awayTeam']['score'])
        home_team_score.append(skater_detail_data[i]['topShotSpeed']['overlay']['homeTeam']['score'])
        period.append(skater_detail_data[i]['topShotSpeed']['overlay']['periodDescriptor']['number'])
        time_in_period.append(skater_detail_data[i]['topShotSpeed']['overlay']['timeInPeriod'])

    except KeyError: 
        max_speed_imp.append('NaN')
        max_speed_met.append('NaN')
        max_speed_percentile.append('NaN')
        game_date.append('NaN')
        away_team.append('NaN')
        home_team.append('NaN')
        away_team_score.append('NaN')
        home_team_score.append('NaN')
        period.append('NaN')
        time_in_period.append('NaN')        
```

```python
#create a data frame with max speed shot data 
max_shot_speed = pd.DataFrame(list(zip(skater_id, shooter_team, max_speed_imp, max_speed_met, max_speed_percentile, game_date,
                                        away_team, home_team, away_team_score, home_team_score, period, time_in_period)),
               columns =['skater_id', 'shooter_team', 'max_speed_imp', 'max_speed_met', 'ax_speed_percentile', 'game_date',
                                        'away_team', 'home_team', 'way_team_score', 'home_team_score', 'period', 'time_in_period'])
```
