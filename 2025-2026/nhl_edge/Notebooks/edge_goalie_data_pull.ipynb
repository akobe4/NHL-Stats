# Goalie Data

### Goalie Detail 

###### Retrieve goalie rankings for NHL Edge data, Includes GAA, games above .900, goal differential per 60, average goal support, point percentage, shot location summary/details.

```python
#create url's for each player id
all_url = []

for i in range(len(goalie_id_list)):
    goalie_id = goalie_id_list[i]
    goalie_id_url = 'https://api-web.nhle.com/v1/edge/goalie-detail/'+str(goalie_id)+'/20252026/2'
    all_url.append(goalie_id_url)
```

```python
# determining which goalie's data are not showing up in data
goalie_detail = []

for i in range(len(all_url)):
    try:
        response = requests.get(all_url[i])
        goalie_detail.append(response.json())
    except JSONDecodeError:
        print(all_url[i])
        continue
```
